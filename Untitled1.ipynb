```python
import geopandas as gpd
import matplotlib.pyplot as plt

alc = gpd.read_file("/content/ALC_Grades_(Provisional)___ADAS_&_Defra.shp")
alc = alc.to_crs("EPSG:27700")
```

```python
alc
```

```python
def classify(ALC_GRADE):
    if ALC_GRADE in ["Grade 1"]:
        return "Low suitability"
    elif ALC_GRADE in ["Grade 3", "Grade 2"]:
        return "Medium suitability"
    elif ALC_GRADE in ["Grade 4", "Grade 5"]:
        return "High suitability"
    else:
        return "Unknown"

alc["suitability"] = alc["ALC_GRADE"].apply(classify)
```

```python
color_map = {
    "High suitability": "green",
    "Medium suitability": "yellow",
    "Low suitability": "red",
    "Unknown": "lightgray"
}

alc["color"] = alc["suitability"].map(color_map)
```

```python
fig, ax = plt.subplots(figsize=(10, 8))

alc.plot(color=alc["color"], ax=ax, edgecolor="black", linewidth=0.2)


import matplotlib.patches as mpatches

legend_patches = [
    mpatches.Patch(color="green", label="High suitability"),
    mpatches.Patch(color="yellow", label="Medium suitability"),
    mpatches.Patch(color="red", label="Low suitability")
]

ax.legend(handles=legend_patches, loc="lower left")


ax.set_title("Agricultural Land Suitability for Floodplain Reconnection")


ax.set_axis_off()

plt.show()
```

```python
boundary = gpd.read_file("/content/NeneRiverBase.shp")

boundary = boundary.set_crs(epsg=27700)
boundary = boundary.to_crs(alc.crs)


boundary["geometry"] = boundary.buffer(0)
alc["geometry"] = alc.buffer(0)


boundary = boundary[boundary.is_valid]
alc = alc[alc.is_valid]


boundary = boundary.dissolve()

clipped = gpd.clip(alc, boundary)
```
